## Task 1 — Canny Edge Detection

An **edge** is where brightness changes sharply. Canny finds edges in 5 steps: Gaussian blur → gradients → non-maximum suppression (thin the edges) → double thresholding → hysteresis (keep weak edges only if they connect to strong ones).

`cv2.Canny(img, T_low, T_high)`:
- `T_low` = lower threshold (weak edges)
- `T_high` = upper threshold (strong edges)
- Rule of thumb: `T_high ≈ 2×T_low` or `3×T_low`.

```python
# --- Code from the lab manual ---
img = data.camera()

edges = cv2.Canny(img, 100, 200) # 3 (image ,lower threshold , ubber) اقل من ال100 يصير 0 مو ايدج
#الاكبر يصير ايدج ويتحول الى 255
# اللي بينهم يصير نفس اللي جنبه اذا متصل ب ايدج يصير ايدج واذا ما كان متصل في ادج يصير 0

plt.subplot(121), plt.imshow(img, cmap='gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(edges, cmap='gray')
plt.title('Edge Image'), plt.xticks([]), plt.yticks([])
plt.show()
```

 ## Task 2 — Harris Corner Detection يركز على الزوايا  لازم احركه جركة كبيره عشان اكتشف وجود الادج

A **corner** is a point where the image changes in *two* different directions, which makes corners highly distinctive.



- Large positive `R` → corner
- Large negative `R` → edge
- Small `R` → flat region

Parameters: `blockSize` (window size), `ksize` (Sobel kernel size), `k` (sensitivity, typically 0.04–0.06).

> **Note:** OpenCV uses BGR order, so `(0, 0, 255)` is red.

```python
# --- Code from the lab manual ---
board = data.checkerboard() # load
image = cv2.cvtColor(board, cv2.COLOR_GRAY2BGR) # blue Grean red خطوه احتياريه
image = cv2.resize(image, (400, 400), interpolation=cv2.INTER_NEAREST) # اختياريه

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) # عشان نسوي معادلات نحولها االى قري سكيل
gray = np.float32(gray) #حولناها الى فلوت لانها كانت انتجر وسبب التحويل استخدام عمليات حسابيه بدون ايرور

# Harris Corner Detection
harris = cv2.cornerHarris(gray, blockSize=2, ksize=3, k=0.04) #4 parametr (image , قد ايش يمشي الخطوه و )
# k قيمه ثابته وافضل شي تكون ,4 الى ,6
# Threshold for detecting strong corners
threshold = 0.01 * harris.max() # اعلى قيم للكورنر  واقواها
corner_coords = np.argwhere(harris > threshold) # اقل لا تعتبره كورنر

# Draw larger red dots (radius = 7)
for y, x in corner_coords:
    cv2.circle(image, (x, y), radius=7, color=(0, 0, 255), thickness=-1)  # -1 = filled , img , x,y ,rad , color , يعني يخليها فيلد سيركل

# Display
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Harris Corners with Bigger Red Dots')
plt.axis('off')
plt.show()
```

```python
# --- Code from the lab manual ---
board = data.checkerboard() # load
image = cv2.cvtColor(board, cv2.COLOR_GRAY2BGR) # blue Grean red خطوه احتياريه
image = cv2.resize(image, (400, 400), interpolation=cv2.INTER_NEAREST) # اختياريه

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) # عشان نسوي معادلات نحولها االى قري سكيل
gray = np.float32(gray) #حولناها الى فلوت لانها كانت انتجر وسبب التحويل استخدام عمليات حسابيه بدون ايرور

# Harris Corner Detection
harris = cv2.cornerHarris(gray, blockSize=2, ksize=3, k=0.04)

# Threshold for detecting strong corners
threshold = 0.01 * harris.max()
corner_coords = np.argwhere(harris > threshold)

# Draw larger red dots (radius = 7)
for y, x in corner_coords:
    cv2.circle(image, (x, y), radius=7, color=(0, 0, 255), thickness=-1)  # -1 = filled

# Display
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Harris Corners with Bigger Red Dots')
plt.axis('off')
plt.show()
```

## Q1 — Otsu's Thresholding (Assessment Task)
Otsu picks `T` **automatically** by maximizing the between-class variance — assumes the histogram has two peaks (foreground + background).

show the original image, its histogram, and the thresholded image.

```python
image = data.camera()

T, thresholded = cv2.threshold(
    image,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.hist(image.ravel(), bins=256, range=(0, 256))
plt.axvline(T, linestyle='--', label=f'Otsu Threshold = {T:.0f}')
plt.title('Image Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')
plt.legend()

plt.subplot(1, 3, 3)
plt.imshow(thresholded, cmap='gray')
plt.title('Otsu Thresholded Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
